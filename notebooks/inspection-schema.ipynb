```python
import pandas as pd
from pathlib import Path
```

```python
csv_dir = Path("../data/raw")

for file in sorted(csv_dir.glob("*.csv")):
    df = pd.read_csv(file, low_memory=False)
    col0 = df.columns[0]
    print(f"{file.name} | rows: {len(df):,} | {col0} unique: {df[col0].is_unique}")
    print("column names:", list(df.columns), "\n")
```

```python
apps = pd.read_csv("../data/raw/applications.csv", low_memory=False)
app_genres = pd.read_csv("../data/raw/application_genres.csv", low_memory=False)

print("Missing values in applications.csv")
print(apps.isna().mean().round(2))

app_genres_merged = apps.merge(app_genres, on = "appid", how = "left")

print(f"Apps before joining:    {len(apps):,}\n \
vs after Joining:      {len(app_genres_merged):,}")
```
